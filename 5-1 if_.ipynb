# 제어문 / 조건문(if문)

```python
money = True

if money:
    print("택시를 타고 가라")
else:
    print("걸어 가라")
```

```python
money = False

if money:
    print("택시를 타고 가라")
else:
    print("걸어 가라")
```

```python
# 풀어쓰기 잘못한 경우
money = True

if money:
    print('택시를')
    print('타고')
    print('가라')
```

## 3. 다양한 조건을 판단하는 elif

```python
# if/else로만 표현
pocket = ['paper', 'cellphone']
card = True

if 'money' in pocket:
    print('택시를 타고 가라')
else:
    if card:
        print('택시를 타고 가라')
    else:
        print('걸어 가라')
```

```python
# elif를 사용
if 'money' in pocket:
    print('택시를 타고 가라')
elif card:
    print('택시를 타고 가라')
else:
    print('걸어 가라')
```

```python
# if문 한 줄로 작성
if 'money' in pocket: pass
else: print('카드를 꺼내라')
```

## 4. 조건부 표현식

```python
#if문
score = 50

if score >= 60:
    message = 'success'
else:
    message = 'failure'

print(message)
```

```python
score = 90
message = 'success' if score >=60 else 'failure'
message
```

```python
# 조건부 표현식(conditional expression)
score = 90
message = 'success' if score >=60 else 'failure'
message
```
