## 7. 문자열 포매팅

```python
# 1) 숫자 바로 대입
food = "i eat %d apples."%5
food
```

```python
# 2) 문자열 바로 대입
food = "i eat %s apples."%'three' 
food
```

```python
# 3) 숫자 값을 나타내는 변수로 대입
number = 3
food = "i eat %d apples."%number
food
```

```python
# 4) 2개 이상의 값 넣기
num = 10
day = 'three'

'I eat %d apples. so i was sick for %s days.'%(num, day)
```

```python
# %s는 숫자도 문자열로 대입 가능
'I have %s apples.'%3.14
```

```python
# 포맷 코드와 문자%를 같이 사용하려면 %%
'Error is %d%%.' %98
```

## 1. format()함수를 사용한 포매팅

```python
# 숫자 바로 대입
'l eat {} apples.' .format(3)
```

```python
# 문자 바로 대입
'l eat {} apples.' .format('three')
```

```python
# 변수로 대입
num = 3
'l eat {} apples.' .format(num)
```

```python
'l eat {0} apples.' .format('three')
```

```python
# 2개 이상 값 넣기
num = 10
day = 'three'
```

```python
'I eat {} apples. so i was sick for {} days.'.format(num, day)
```

```python
'I eat {1} apples. so i was sick for {0} days.'.format(num, day)
```

```python
# 이름으로 넣기
'I eat {number} apples. so i was sick for {day} days.'.format(day=3, number=10)
```

```python
# 왼쪽 정렬
'{}'.format('hi')
```

```python
'{:<10}'.format('hi')
```

```python
'{:>10}'.format('hi')
```

```python
'{:^10}'.format('hi')
```

```python
# 공백 채우기
'{:=^10}'.format('hi')
```

```python
'{:!<10}'.format('hi')
```

```python
# 소수점 표현
'{:0.4f}'.format(3.42789798)
```

```python
'{:10.4f}'.format(3.42789798)
```

```python
'{:0.2f}'.format(3.42789798)
```

```python
'{{and}}'.format()
```

## 11. f문자열 포매팅

```python
# 문자열 앞에 f접두사를 붙이기
name = '정민'
age = 20
f'나의 이름은 {name}입니다. 나이는 {age}입니다.'
```

```python
# 표현식 사용하기
f'나는 내년이면 {age + 1}살이 됩니다.'
```

```python
d = {'name' : '정민', 'age' : 20}
f'나의 이름은 {name}입니다. 나이는 {age}입니다.'
```

```python
d = {'name' : '정민', 'age' : 20}
f'나의 이름은 {d['name']}입니다. 나이는 {d['age']}입니다.'
```

```python
# 정렬
f'{'hi':<10}'
```

```python
f'{'hi':^10}'
```

```python
f'{'hi':!<10}'
```

```python
# 소수점 표현
f'{3.44356536:0.2f}'
```

```python
f'{3.44356536:10.2f}'
```

```python
f'{3.44356536:0<10.2f}'
```

```python
f'{{and}}'
```
